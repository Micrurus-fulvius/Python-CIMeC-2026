## Exercise 8 — Handle two different exception types

Complete the code so that it:
- catches `ZeroDivisionError` and prints `"Division by zero is not allowed."`
- catches `TypeError` and prints `"Incompatible types for division."`

Test it with:
- `a = 10, b = 0`
- `a = "10", b = 2`

````python
a = 10
b = 0

try:
    result = a / b
    print(result)
except ZeroDivisionError:
    print("Division by zero is not allowed.")
except TypeError:
    print("Incompatible types for division.")
````

## Exercise 10 — Debugging workflow

The following code contains ** errors**.  
Run it, read the traceback, fix the error, and run again until the full code works.

````python

from math import squareroot

values = [4, 9, 16]
results = []

for i in range(4):
    results.append(squareroot(values[i]))

print("Square roots: " + results)
````


Questions:
1. Which error do you get first? ImportError (si chiama sqrt)
2. After fixing it, which error appears next? NameError; poi IndexError
3. What is the final working version of the code? vedi sotto

````python

from math import sqrt

values = [4, 9, 16]
results = []

for i in range(len(values)):
    results.append(sqrt(values[i]))

print("Square roots: " + str(results))
````
